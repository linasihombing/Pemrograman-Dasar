**Soal 1. Daya Generator dan Efisiensi Turbin (10 poin | CLO-1 | TP-1, TP-3)**

Sebuah generator menghasilkan tegangan 220 Volt dan arus 15 Ampere. Sebuah turbin menerima energi_input = 500 dan menghasilkan energi_output = 400.

a. Simpan nilai tegangan dan arus dalam variabel, hitung daya listrik (P = V × I), lalu tampilkan dengan f-string. b. Hitung efisiensi turbin: Efisiensi (%) = (energi_output / energi_input) × 100. Tampilkan dengan 2 angka desimal.

Contoh output:

Daya listrik = 3300 W

Efisiensi = 80.00%

```python
# Soal 1. Daya Generator dan Efisiensi Turbin

# a. Daya listrik

tegangan = 220          # Volt
arus = 15               # Ampere
daya = tegangan * arus  # P = V x I
print(f"Daya listrik = {daya} W")
```

```python
# b. Efisiensi turbin

energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100
print(f"Efisiensi = {efisiensi:.2f}%")
```

**Soal 5. Katup Pengaman (20 poin | CLO-1 | TP-3, TP-4)**

Status sebuah katup pengaman ditentukan sebagai berikut:

"Katup Terbuka" jika tekanan_uap lebih dari 150 psi dan status_manual_override bernilai False.
"Katup Mode Manual" jika status_manual_override bernilai True.

selain itu, "Katup Tertutup".

Baca tekanan_uap dari input (bertipe float). Nilai status_manual_override diberikan langsung dalam kode, misalnya status_manual_override = False. Gunakan if-elif-else dengan operator logika.

Uji program dengan kombinasi berikut:
Screenshot 2026-11-10 114815.png

```python
# Soal 5. Katup Pengaman

tekanan_uap = float(input("Masukkan tekanan uap (psi): "))
status_manual_override = False   # ubah ke True untuk menguji mode manual

if tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")
elif status_manual_override == True:
    print("Katup Mode Manual")
else:
    print("Katup Tertutup")
```

Cara kerja program:

1. Program meminta pengguna mengetik tekanan uap.
2. Status override sudah ditentukan di kode (False).
3. Python memeriksa kondisi dari atas ke bawah, lalu menjalankan satu cabang pertama yang benar.
4. Karena hanya ada satu pasang nilai, outputnya hanya satu baris.

Kode Fungsi :

input("Masukkan tekanan uap (psi): ")	untuk menampilkan pertanyaan dan membaca ketikan pengguna sebagai teks.

float(...)	untuk mengubah teks menjadi angka desimal agar bisa dibandingkan dengan 150.

tekanan_uap = ...	untuk menyimpan angka hasil konversi ke variabel.

if tekanan_uap > 150 and status_manual_override == False:	Benar hanya jika tekanan lebih dari 150 dan override sama dengan False. == adalah operator perbandingan, and mengharuskan kedua syarat benar.

print("Katup Terbuka")	Dijalankan jika kondisi if benar.

elif status_manual_override == True:	Diperiksa hanya jika if salah. Benar jika override aktif.

print("Katup Mode Manual")	Dijalankan jika kondisi elif benar.

else:	Dijalankan jika semua kondisi di atas salah.

print("Katup Tertutup")	Output untuk kondisi selain di atas.

Namun, untuk menguji tekanan uap lain, harus mengubah False/True secara manual lalu menjalankan ulang sel. Maka dari itu, untuk menguji tekanan uap secara keseluruhan dapat menggunakan program berikut ini :

```python
# Menguji semua tekanan uap secara keseluruhan

data_uji = [(160, False), (160, True), (140, False), (150, False), (140, True)]

for tekanan_uap, status_manual_override in data_uji:
    if tekanan_uap > 150 and status_manual_override == False:
        hasil = "Katup Terbuka"
    elif status_manual_override == True:
        hasil = "Katup Mode Manual"
    else:
        hasil = "Katup Tertutup"
    print(f"{tekanan_uap} -> {status_manual_override} -> {hasil}")
```

Cara kerja program:

1. Kelima kombinasi dari tabel soal disimpan dalam satu list.
2. Loop for mengambil satu kombinasi per putaran, lalu memasukkan nilainya ke dua variabel.
3. Kondisi if-elif-else yang sama dijalankan untuk tiap kombinasi dan hasilnya disimpan di hasil.
4. print menampilkan kombinasi beserta hasilnya, sehingga muncul 5 baris output.

Kode	Fungsi :

if ... and ... == False:	Logika yang sama dengan versi 1: tekanan > 150 dan override False.

elif status_manual_override == True:	Jika override aktif, hasilnya mode manual.

else:	Selain itu, katup tertutup.

**Soal 7. Pemanas Mencapai Target (25 poin | CLO-1 | TP-5)**

Sebuah sistem pemanas menaikkan suhu hingga target 85 °C. Mulai dengan suhu_saat_ini = 60. Gunakan perulangan while: selama suhu_saat_ini < 85, naikkan 5 derajat, lalu cetak suhu saat ini (setelah dinaikkan). Saat suhu mencapai tepat 80, cetak "WARNING: Mendekati target!". Setelah loop selesai, cetak pesan target tercapai.

```python
# Soal 7. Pemanas Mencapai Target

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")
    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

**Soal 9. Pembacaan Vibrasi dengan Sensor Rusak (30 poin | CLO-1 | TP-5, TP-6)**

Data vibrasi lima unit mesin dibaca sebagai teks, dan satu sensor rusak mengirim "ERR":

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

Periksa setiap data dengan for. Nomor unit dimulai dari 1. a. Ubah data menjadi float dengan try-except ValueError. Jika gagal, cetak Unit [n]: data tidak valid, tambahkan ke penghitung data tidak valid, lalu lanjutkan ke unit berikutnya. b. Jika nilai melebihi 2.0, cetak Warning: Unit [n] vibrasi tinggi!. c. Jika nilai melebihi 3.0, setelah mencetak Warning, cetak KRITIS: pemeriksaan dihentikan lalu hentikan perulangan dengan break. d. Setelah perulangan selesai, tampilkan jumlah data tidak valid.

```python
# Soal 9. Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for n, teks in enumerate(data, start=1):
# a. Konversi ke float dengan try-except
    try:
        nilai = float(teks)
    except ValueError:
        print("(a)", f"Unit {n}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

# b. Vibrasi tinggi
    if nilai > 2.0:
        print("(b)", f"Warning: Unit {n} vibrasi tinggi!")

# c. Kondisi kritis
    if nilai > 3.0:
            print("(c)", "KRITIS: pemeriksaan dihentikan")
            break

# d. Jumlah data tidak valid
print("(d)", f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

**Challenge: Bonus Nilai (hingga +25 poin) (CLO-1 | TP-4, TP-5, TP-6)**

Buatlah program yang menyelesaikan satu masalah di bidang energi (mis. pemantauan beban harian, deteksi anomali suhu, simulasi pengisian/pengosongan, atau penghitungan konsumsi listrik).

Program minimal memuat:

percabangan if-elif-else,
perulangan for atau while,
sebuah list untuk menyimpan data.
atau dengan kombinasi lainya.
Tambahkan komentar yang menjelaskan masalah dan cara kerja program. Penggunaan try-except dan variabel/ekspresi yang tepat akan dinilai lebih baik.

```python
# CHALLENGE: Simulasi Pengisian & Pengosongan Baterai

# MASALAH:
# Sebuah baterai penyimpanan enertgi (kapasitas 10 KWh) mengalami perubahan
# level tiap jam akibat oengisian (+) dan pemakaian (-). Data perubahan level
# (dalam %) dikirm sensor sebagai tek, dan kadang sensor rusak sehinggga
# mengirim "ERR"

# CARA KERJA:
# 1. Perubahan level per jam disimpan dalam list (bertipe teks).
# 2. Perulangan for membaca data jam demi jam.
# 3. try-except ValueError menangani data rusak (dilewati).
# 4. Level baterai diperbarui dan dibatasi antara 0 - 100 %.
# 5. if-elif-else menentukan status: PENUH, NORMAL, RENDAH, KRITIS.
# 6. Jika level KRITIS, simulasi dihentikan dengan break.
# 7. Setelah loop, ditampilkan ringkasan hasil simulasi.


# Data & parameter
perubahan_level = ["+20", "+35", "-10", "ERR", "+10", "-30", "-25", "+40", "-60"]

KAPASITAS_KWH = 10      # kapasitas baterai (kWh)
level = 50.0            # level awal (%)

riwayat_level = []      # list untuk menyimpan level setiap jam valid
jumlah_tidak_valid = 0
jumlah_penuh = 0
dihentikan = False

print("=== SIMULASI BATERAI ===")
print(f"Level awal: {level:.1f}%\n")

# Perulangan simulasi
for jam, teks in enumerate(perubahan_level, start=1):
    # Konversi teks ke float, tangani sensor rusak
    try:
        delta = float(teks)
    except ValueError:
        print(f"Jam {jam}: data tidak valid ('{teks}'), dilewati")
        jumlah_tidak_valid += 1
        continue

    # Perbarui level dan batasi 0 - 100 %
    level = level + delta
    if level > 100:
        level = 100.0
    elif level < 0:
        level = 0.0

    riwayat_level.append(level)

    # Tentukan status baterai
    if level >= 100:
        status = "PENUH"
        jumlah_penuh += 1
    elif level >= 50:
        status = "NORMAL"
    elif level > 10:
        status = "RENDAH"
    else:
        status = "KRITIS"

    print(f"Jam {jam}: {teks:>4} -> level {level:5.1f}% | {status}")

    # Hentikan simulasi jika kritis
    if status == "KRITIS":
        print("KRITIS: baterai hampir habis, simulasi dihentikan!")
        dihentikan = True
        break

# Ringkasan
print("\n=== RINGKASAN ===")

if len(riwayat_level) > 0:
    level_akhir = riwayat_level[-1]
    energi_sisa = level_akhir / 100 * KAPASITAS_KWH

    print(f"Jumlah data valid      : {len(riwayat_level)}")
    print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
    print(f"Level tertinggi        : {max(riwayat_level):.1f}%")
    print(f"Level terendah         : {min(riwayat_level):.1f}%")
    print(f"Jumlah jam status PENUH: {jumlah_penuh}")
    print(f"Level akhir            : {level_akhir:.1f}%")
    print(f"Energi tersisa         : {energi_sisa:.2f} kWh")
else:
    print("Tidak ada data valid.")

if dihentikan:
    print("Catatan: simulasi berhenti karena baterai kritis.")
else:
    print("Catatan: seluruh data berhasil diproses.")
```
