1. Buatlah program Python yang meminta pengguna memasukkan dua bilangan bulat, lalu tampilkan hasil dari operasi +, -, *, /, // (pembagian bulat), dan % (modulus) antara kedua bilangan tersebut.

```python
# Nama    : Maulina Sihombing
# Nim     : 4232501058
# Tanggal : 24/09/2026
# Tujuan  : Program ini bertujuan untuk mengimplementasikan dan menampilkan hasil penggunaan operator aritmatika dasar hingga spesifik dalam bahasa pemrograman
# Python. Melalui pengerjaan program ini, diharapkan dapat memahami cara kerja serta perbedaan dari operasi penjumlahan, pengurangan, perkalian, pembagian
# riil (/), pembagian bulat (//), dan modulus (%) terhadap dua input bilangan bulat.

# JAWABAN SOAL UMUM NO.1

bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

# Menampilkan hasil operasi
print("\n" + "Bilangan pertama:", bilangan1)
print("Bilangan kedua:", bilangan2)

print()
print("Hasil operasi:")
print("Penjumlahan (+)      :", bilangan1 + bilangan2)
print("Pengurangan (-)      :", bilangan1 - bilangan2)
print("Perkalian (*)        :", bilangan1 * bilangan2)
print("Pembagian (/)        :", bilangan1 / bilangan2)
print("Pembagian bulat (//) :", bilangan1 // bilangan2)
print("Modulus (%)          :", bilangan1 % bilangan2)
```

3. Buatlah program yang meminta pengguna memasukkan umur dan status memiliki SIM (True/False), lalu gunakan operator logika (and) untuk menentukan apakah orang tersebut "boleh mengemudi sendiri" (syarat: umur ≥ 17 dan punya SIM). Tambahkan minimal satu ekspresi yang sengaja butuh tanda kurung agar hasilnya benar, sebagai latihan operator precedence.

```python
# Nama    : Maulina Sihombing
# Nim     : 4232501058
# Tanggal : 24/09/2026
# Tujuan  : Program ini bertujuan untuk menentukan kelayakan seseorang mengemudimsecara mandiri dengan memanfaatkan operator logika dan mempraktikkan konsep
# urutan eksekusi operator (operator precedence). Melalui pengerjaan program ini, pengguna dapat memahami cara menggabungkan beberapa kondisi syarat (umur
# 17 tahun dan kepemilikan SIM) menggunakan operator logika and, mengolah input boolean dari teks, serta mengontrol prioritas evaluasi ekspresi menggunakan
# tanda kurung agar menghasilkan keputusan logika yang tepat.

# JAWABAN SOAL UMUM NO.3
# KEADAAN TRUE

# Meminta input dari pengguna
umur = int(input("Masukkan umur Anda: "))
punya_sim = input("Apakah Anda memiliki SIM? (True/False): ").strip().lower() == 'true'

# Ekspresi dengan tanda kurung untuk latihan operator precedence
# (umur >= 17) and punya_sim
# Tanda kurung memastikan umur >= 17 dievaluasi dulu sebelum and
boleh_mengemudi = (umur >= 17) and punya_sim

# Menampilkan hasil
print("\n" + "="*40)
print("HASIL PENGECEKAN")
print("="*40)
print(f"Umur: {umur} tahun")
print(f"Memiliki SIM: {punya_sim}")
print("-"*40)

if boleh_mengemudi:
    print("Anda BOLEH mengemudi sendiri!")
else:
    print("Anda TIDAK BOLEH mengemudi sendiri!")

# Contoh tambahan dengan operator precedence yang lebih kompleks
print("\n" + "="*40)
print("CONTOH OPERATOR PRECEDENCE")
print("="*40)
print("Ekspresi: (umur >= 17) and punya_sim")
print(f"({umur} >= 17) = {umur >= 17}")
print(f"Hasil akhir: {boleh_mengemudi}")
```

```python
# KEADAAN FALSE
# Menentukan apakah seseorang boleh mengemudi sendiri

# Meminta input dari pengguna
umur = int(input("Masukkan umur Anda: "))
punya_sim = input("Apakah Anda memiliki SIM? (True/False): ").strip().lower() == 'true'

# Ekspresi dengan tanda kurung untuk latihan operator precedence
# (umur >= 17) and punya_sim
# Tanda kurung memastikan umur >= 17 dievaluasi dulu sebelum and
boleh_mengemudi = (umur >= 17) and punya_sim

# Menampilkan hasil
print("\n" + "="*40)
print("HASIL PENGECEKAN")
print("="*40)
print(f"Umur: {umur} tahun")
print(f"Memiliki SIM: {punya_sim}")
print("-"*40)

if boleh_mengemudi:
    print("Anda BOLEH mengemudi sendiri!")
else:
    print("Anda TIDAK BOLEH mengemudi sendiri!")

# Contoh tambahan dengan operator precedence yang lebih kompleks
print("\n" + "="*40)
print("CONTOH OPERATOR PRECEDENCE")
print("="*40)
print("Ekspresi: (umur >= 17) and punya_sim")
print(f"({umur} >= 17) = {umur >= 17}")
print(f"Hasil akhir: {boleh_mengemudi}")
```

4. Buatlah program Python yang menghitung efisiensi termal sebuah unit pembangkit menggunakan rumus: efisiensi (%) = (energi output / energi input) * 100

Program meminta pengguna memasukkan nilai energi input dan energi output (dalam MJ), lalu menampilkan hasil efisiensi dengan f-string, dibulatkan 2 angka desimal. Pastikan penggunaan tanda kurung pada rumus sudah benar sesuai operator precedence.

```python
# Nama    : Maulina Sihombing
# Nim     : 4232501058
# Tanggal : 24/09/2026
# Tujuan  : Program ini bertujuan untuk menghitung persentase efisiensi termal unit pembangkit berdasarkan masukan nilai energi input dan output dalam satuan
# Megajoule (MJ). Melalui pengerjaan program ini, pengguna dapat memahami penerapan operator aritmatika pembagian (/) dan perkalian (*) dengan tanda kurung
# eksplisit untuk mengatur operator precedence, pengolahan tipe data desimal (float), serta pemformatan string menggunakan f-string {efisiensi:.2f} untuk
# membulatkan hasil numerik hingga dua angka di belakang koma.

#JAWABAN SOAL UMUM NO.4

# Meminta input energi input dan energi output
energi_input  = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

# Hitung efisiensi dengan tanda kurung eksplisit
efisiensi = (energi_output / energi_input) * 100

# Tampilkan hasil dengan 2 angka desimal
print(f"Efisiensi termal: {efisiensi:.2f}%")
```

5. Buatlah program yang meminta pengguna memasukkan suhu boiler dan beban operasi (%) suatu unit pembangkit, lalu gunakan operator relasional dan logika untuk menghitung apakah unit tersebut dalam kondisi "perlu maintenance" — yaitu benar (True) jika suhu boiler > 550°C atau beban operasi > 95%.
Tampilkan hasilnya (nilai True/False) dalam satu kalimat naratif menggunakan f-string, tanpa menggunakan if/else

```python
# Nama    : Maulina Sihombing
# Nim     : 4232501058
# Tanggal : 24/09/2026
# Tujuan  : Program ini bertujuan untuk mengevaluasi status pemeliharaan (maintenance) unit pembangkit berdasarkan ambang batas suhu boiler dan beban
# operasi menggunakan gabungan operator relasional dan operator logika. Melalui pengerjaan program ini, pengguna dapat memahami cara menerapkan operator
# logika or untuk memeriksa apakah salah satu kondisi terpenuhi (suhu boiler >550 derajat celciius atau beban operasi > 95%), serta menampilkan output
# berupa nilai kebenaran boolean secara langsung ke dalam kalimat naratif menggunakan f-string tanpa perlu mengandalkan struktur percabangan (if/else).

# JAWABAN SOAL UMUM NO.5
# Program untuk mengecek status maintenance unit pembangkit
# Keadaan False

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = suhu_boiler > 550 or beban_operasi > 95

print(f"Dengan suhu boiler {suhu_boiler}°C dan beban operasi {beban_operasi}%, "
      f"status 'perlu maintenance' untuk unit ini adalah {perlu_maintenance}.")
```
