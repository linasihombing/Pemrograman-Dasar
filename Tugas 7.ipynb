```python
# SOAL L1 - Pola Angka Segitiga

print("--- Soal L1: Pola Angka ---")
for baris in range(1, 6):                 # Outer loop: baris 1 s.d. 5
    for angka in range(1, baris + 1):     # Inner loop: cetak 1 s.d. nomor baris
        print(angka, end=" ")
    print()    # Pindah baris
```

```python
# SOAL L2 - Tabel Data Sensor Pembangkit

data_daya = [
    [142, 138, 145, 140],    # Unit A
    [115, 120, 118, 122],    # Unit B
    [ 95,  98,  92,  96],    # Unit C
]
nama_unit = ["Unit A", "Unit B", "Unit C"]

print("--- Soal L2: Rata-rata Daya per Unit ---")
for i in range(len(data_daya)):                 # iterasi setiap unit
    total = 0
    for daya in data_daya[i]:                   # iterasi setiap hari pada unit ke-i
        total += daya
    rata2 = total / len(data_daya[i])           # bagi dengan jumlah hari
    print(f"{nama_unit[i]}: {rata2:.1f} MW")
```

```python
# SOAL L3 - Gabungan while + Nested Loop

print("--- Soal L3: Monitoring Sensor ---")
import random
random.seed(42)

siklus = 0
ada_anomali = True

while ada_anomali:
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    for sensor_id in range(1, 5):
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:
            print(f" ⚠ Sensor {sensor_id}: {suhu}°C - ANOMALI")
            ada_anomali = True
        else:
            print(f" ✓ Sensor {sensor_id}: {suhu}°C - NORMAL")

print(f"\n Semua sensor normal pada siklus ke-{siklus}.")
```
