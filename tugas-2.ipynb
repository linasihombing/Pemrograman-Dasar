**Latihan 4 — List (sedang)**

Berikut daftar pembacaan tegangan (V) sebuah panel:

```python
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]
```

Tugas: (a) cetak pembacaan pertama dan terakhir memakai indeks, (b) tambahkan satu pembacaan baru 222.0 dengan append(), (c) cetak jumlah total data dengan len().

```python
#Jawaban latihan 4

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

#A. Cetak pembacaan pertama dan terakhir
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[5])

#B. Tambahkan pembacaan baru 222.0
tegangan_harian.append(222.0)
print("tegangan harian:", tegangan_harian)

#C. Cetak jumlah total data
print("Jumlah total data:", len(tegangan_harian))
```

**Latihan 5 — Dictionary (sedang)**

Buat dictionary spesifikasi sebuah motor listrik (minimal: nama, daya_kW, tegangan_V, rpm). Cetak salah satu nilai lewat key-nya, lalu ubah nilai rpm (pakai =), dan cetak ulang seluruh dictionary.

```python
#Jawaban latihan 5

#Dictionary spesifikasi
motor_listrik = {
"nama": "Motor Induksi 3 Fasa",
"daya_kW": 15.0,
"tegangan_V": 380,
"rpm": 1450
}
print("nama:", motor_listrik["nama"],  "daya:", motor_listrik["daya_kW"],"KW", "tegangan:", motor_listrik["tegangan_V"],"V", "rpm:", motor_listrik["rpm"])

#Cetak salah satu nilai lewat key
print("tegangan:", motor_listrik["tegangan_V"], "V")

#Ubah nilai rpm
motor_listrik["rpm"] = 1500

#Cetak ulang seluruh dictionary
print("nama:", motor_listrik["nama"],  "daya:", motor_listrik["daya_kW"],"KW", "tegangan:", motor_listrik["tegangan_V"],"V", "rpm:", motor_listrik["rpm"])
```

Latihan 6 — Gabungan (menantang)

Kelola data tiga unit pembangkit sebagai list of dictionary, tiap unit berisi nama, kapasitas_MW, dan status. Tugas: cetak nama unit kedua, ubah status salah satu unit, tambahkan satu unit baru ke list dengan append(), lalu cetak seluruh data.

```python
#Jawaban Latihan 6
pembangkit = [
{"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
{"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
{"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

#A.cetak nama unit kedua
print("nama:", pembangkit[1]["nama"])

#B.ubah status salah satu unit (misal Unit-3 menjadi "operasi")
pembangkit[2]["status"] = "operasi"
print("status:", pembangkit[2]["status"])

#C.tambahkan unit baru ke list dengan append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 200, "status": "operasi"}
pembangkit.append(unit_baru)


#D.cetak seluruh data
print("1.", pembangkit[0]["nama"], "-", pembangkit[0]["kapasitas_MW"], "MW -", pembangkit[0]["status"])
print("2.", pembangkit[1]["nama"], "-", pembangkit[1]["kapasitas_MW"], "MW -", pembangkit[1]["status"])
print("3.", pembangkit[2]["nama"], "-", pembangkit[2]["kapasitas_MW"], "MW -", pembangkit[2]["status"])
print("4.", pembangkit[3]["nama"], "-", pembangkit[3]["kapasitas_MW"], "MW -", pembangkit[3]["status"])
```
