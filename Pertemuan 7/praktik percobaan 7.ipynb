# **Praktikum 1: Perulangan for dengan range()**


```python
# === PRAKTIKUM 1A: Bilangan Ganjil 1–9 ===
print("--- Bilangan Ganjil 1 hingga 9 ---")
for angka in range(1, 10, 2):   # start=1, stop=10, step=2
    print(angka)
```

```python
# === PRAKTIKUM 1B: Bilangan Genap 2 hingga 50 ===
print("--- Bilangan Genap 2 hingga 50 ---")
for angka in range(2, 51, 2):
    print(angka, end=" ")
print()  # pindah baris setelah selesai
```

```python
# === PRAKTIKUM 1C: Hitung Mundur dari 255 ===
print("--- Countdown dari 255 ke 0 ---")
for i in range(255, -1, -1):    # step=-1 untuk mundur
    print(i, end=" ")
print()
```

# **Praktikum 2: Perulangan while dan Nilai Penghenti**

```python
# === PRAKTIKUM 2A: while Dasar ===
hitung = 1
while hitung <= 5:
    print(f"Iterasi ke-{hitung}")
    hitung += 1  # sangat penting: tanpa ini, loop tak akan berhenti
print("Perulangan selesai.")
```

```python
# === PRAKTIKUM 2A: while Dasar ===
hitung = 1
while hitung <= 5:
    print(f"Iterasi ke-{hitung}")
print("Perulangan selesai.")
```

```python
# === PRAKTIKUM 2B: Nilai Penghenti ===
kode_berhenti = 99
input_pengguna = 0

while input_pengguna != kode_berhenti:
    try:
        input_pengguna = int(input("Masukkan angka (99 untuk berhenti): "))
        if input_pengguna != kode_berhenti:
            print(f"  → Anda memasukkan: {input_pengguna}")
    except ValueError:
        print("  Input tidak valid. Harap masukkan angka.")

print("Perulangan telah dihentikan oleh Nilai Penghenti (99).")
```

```python
# === PRAKTIKUM 2C: Nilai Penghenti dengan NIM ===
NIM = 4232501058  # ganti dengan NIM Anda
input_pengguna = 0

while input_pengguna != NIM:
    try:
        input_pengguna = int(input("Masukkan NIM untuk berhenti: "))
        if input_pengguna != NIM:
            print("  NIM belum cocok, coba lagi.")
    except ValueError:
        print("  Input tidak valid.")

print(f"NIM {NIM} cocok. Program berhenti.")
```

```python
# === PRAKTIKUM 2D: Batas Percobaan Password ===
percobaan_maksimal = 3
percobaan_saat_ini = 1
password_benar = "polibatam"

while percobaan_saat_ini <= percobaan_maksimal:
    masukan = input(f"Percobaan ke-{percobaan_saat_ini}: Masukkan password: ")
    if masukan == password_benar:
        print("Akses diterima!")
        break
    else:
        print("  Password salah.")
    percobaan_saat_ini += 1
else:
    print("Kesempatan habis. Akses ditolak.")
```

PERTANYAAN ANLISIS

1. Apa bedanya while input != 99 dengan for i in range(99)?
    
    Perbedaan utama antara `while input != 99` dan `for i in range(99)` terletak pada cara keduanya menentukan kapan perulangan berhenti. Pada `while input != 99`, perulangan dikendalikan oleh sebuah kondisi, sehingga jumlah iterasinya tidak pasti dan bergantung pada kapan kondisi tersebut tidak lagi terpenuhi. Dalam kasus ini, perulangan terus berjalan sampai pengguna memasukkan angka 99, yang berperan sebagai nilai penghenti. Bisa saja pengguna langsung memasukkan 99 pada percobaan pertama, atau baru memasukkannya setelah puluhan kali mencoba.

    Sebaliknya, `for i in range(99)` dikendalikan oleh urutan angka yang dihasilkan oleh `range()`, sehingga jumlah iterasinya sudah pasti, yaitu tepat 99 kali dengan nilai `i` dari 0 sampai 98. Angka 99 di sini hanya berfungsi sebagai batas atas yang tidak ikut dihitung, bukan sebagai sinyal untuk berhenti.

    Dengan demikian, `while` lebih cocok digunakan ketika jumlah pengulangan tidak diketahui di awal dan bergantung pada suatu kejadian, seperti input pengguna, sedangkan `for` dengan `range()` lebih cocok digunakan ketika jumlah pengulangan sudah diketahui sebelumnya.

    Fungsi input() selalu mengembalikan string, dan pengguna bisa mengetik apa saja. # **Praktikum 3: Iterasi pada List dan String**

```python
# === PRAKTIKUM 3A: Total Belanja ===
harga_barang = [15000, 25000, 10000, 5000, 8000,
                12000, 30000, 7500, 22000, 18000,
                9500, 4000]   # minimal 10 item
total_belanja = 0

for harga in harga_barang:
    total_belanja += harga

print(f"Jumlah item  : {len(harga_barang)}")
print(f"Total belanja: Rp {total_belanja:,.0f}")
```

```python
# === PRAKTIKUM 3B: Iterasi String ===
nama_institusi = "Politeknik Negeri Batam"
hasil_output = ""

for karakter in nama_institusi:
    hasil_output += karakter + "-"

print(hasil_output)
# Output: P-o-l-i-t-e-k-n-i-k- -N-e-g-e-r-i- -B-a-t-a-m-
```

```python
# === PRAKTIKUM 3C: enumerate() dengan nama bulan ===
bulan = ["Januari", "Februari", "Maret", "April",
         "Mei", "Juni", "Juli", "Agustus",
         "September", "Oktober", "November", "Desember"]

for i, nama in enumerate(bulan):
    print(f"Bulan ke-{i + 1:>2}: {nama}")
```

```python
# === PRAKTIKUM 3D: Kelurahan di Kota Batam ===
kelurahan_batam = [
    "Belian", "Baloi Permai", "Taman Baloi", "Sukajadi",
    "Batu Selicin", "Lubuk Baja Kota", "Kampung Pelita",
    "Batu Merah", "Tanjung Uma", "Sungai Jodoh",
    "Nongsa", "Sambau"
]

print("Daftar Kelurahan di Kota Batam:")
print("-" * 35)
for i, nama in enumerate(kelurahan_batam):
    print(f"{i + 1:>3}. {nama}")
print(f"\nTotal: {len(kelurahan_batam)} kelurahan ditampilkan")
```

PERTANYAAN ANALISIS

1. Apa perbedaan for i in range(len(list)) dengan for item in list? Kapan masing-masing lebih tepat digunakan?

    Pada for item in list, variabel item langsung berisi elemen list pada setiap iterasi, sehingga kodenya lebih ringkas dan mudah dibaca. Pada for i in range(len(list)), variabel i berisi indeks (0, 1, 2, ...), sehingga elemen harus diakses lewat list[i].

    Bentuk for item in list lebih tepat jika hanya nilai elemen yang dibutuhkan, misalnya menjumlahkan harga. Bentuk for i in range(len(list)) lebih tepat jika indeks diperlukan, misalnya untuk mengubah isi list (list[i] = list[i] * 2), membandingkan elemen dengan elemen sebelumnya atau sesudahnya, atau menelusuri dua list sekaligus berdasarkan posisi. Namun jika yang dibutuhkan indeks sekaligus nilai, enumerate() biasanya lebih rapi daripada range(len(...)).

# **Praktikum 4: Kontrol Perulangan: break dan continue**

```python
# === PRAKTIKUM 4A: break saat menemukan angka -1 ===
data_angka = [10, 20, 30, -1, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")

print("Program selesai.")
```

```python
# === PRAKTIKUM 4B: break saat menemukan angka 0 ===
data_sensor = [45.2, 67.8, 89.1, 0, 55.3, 72.4]

print("Membaca data sensor pembangkit:")
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati. Proses dihentikan.")
        break
    print(f"  Suhu terbaca: {nilai}°C")
```

```python
# === PRAKTIKUM 4C: continue untuk melewati angka negatif ===
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]

print("Data daya valid (MW):")
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")
```

```python
# === PRAKTIKUM 4D: Penambahan berkelanjutan sampai total > 20 ===
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
```

PERTANYAAN ANALISIS

1. Apa perbedaan perilaku program pada Langkah 1 jika break diganti dengan continue?

    Jika break diganti continue, loop tidak lagi berhenti saat menemukan -1. Program hanya melewati sisa isi loop untuk angka -1 (yaitu baris print("Memproses: ...")), lalu lanjut ke elemen berikutnya, sehingga 50 dan 60 ikut diproses. Outputnya menjadi:


```python
data_angka = [10, 20, 30, -1, 50, 60]

print("Memproses data:")
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        continue
    print(f"  Memproses: {angka}")

print("Program selesai.")
```

Pesan “Proses dihentikan” tetap tercetak, tetapi sebenarnya menyesatkan karena proses tidak berhenti. Perbedaan intinya: break mengakhiri seluruh perulangan, sedangkan continue hanya melewati iterasi saat ini dan melanjutkan ke iterasi berikutnya.

2. Mengapa continue lebih baik digunakan daripada if-else yang panjang dalam kasus Langkah 3?

    Tanpa continue, kode harus ditulis seperti ini:


```python
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
    else:
        print(f"  Daya: {daya} MW")
        # ...pemrosesan lain yang panjang...
```

Semua logika pemrosesan data valid menjadi menjorok (nested) di dalam else. Jika pemrosesannya panjang atau ada beberapa syarat validasi, kode makin dalam bersarang dan sulit dibaca. Dengan continue, data tidak valid disaring di awal (disebut pola guard clause), sehingga sisa kode berada pada level indentasi yang sama dan alurnya lebih jelas: “jika tidak valid, lewati; selebihnya proses seperti biasa”. Hasilnya kode lebih rata, mudah dibaca, dan lebih mudah dirawat.

3. Pada Langkah 4, nilai berapa yang pertama kali membuat total melebihi 20? Hitung secara manual dan verifikasi dengan program.

```python
# JAWABAN NO. 3
# === PERHITUNGAN MANUAL ===
# Iterasi 1: 0 + 1 = 1   (<= 20, lanjut)
# Iterasi 2: 1 + 2 = 3   (<= 20, lanjut)
# Iterasi 3: 3 + 3 = 6   (<= 20, lanjut)
# Iterasi 4: 6 + 4 = 10  (<= 20, lanjut)
# Iterasi 5: 10 + 5 = 15 (<= 20, lanjut)
# Iterasi 6: 15 + 6 = 21 (> 20, berhenti)
# Kesimpulan: total pertama yang melebihi 20 adalah 21, saat bilangan 6 ditambahkan.

# === VERIFIKASI DENGAN PROGRAM ===
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    print(f"Menambahkan {bilangan} -> total = {total}")
    bilangan += 1

print(f"\nTotal pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
```

Jadi total pertama yang melebihi 20 adalah 21, yang terjadi saat bilangan 6 ditambahkan. Ini cocok dengan rumus jumlah deret 1 + 2 + ... + n = n(n+1)/2: untuk n = 5 hasilnya 15 (belum lebih dari 20), sedangkan untuk n = 6 hasilnya 21.

Verifikasi dengan program: output Total pertama yang melebihi 20: 21 dan Bilangan terakhir ditambahkan: 6. Baris kedua memakai bilangan - 1 karena setelah penambahan, variabel bilangan sudah dinaikkan satu kali lagi menjadi 7, sehingga bilangan terakhir yang benar-benar ditambahkan adalah 6.

# **Praktikum 5: Perulangan Bersarang (Nested Loop)**

```python
# === PRAKTIKUM 5A: Mencetak "Python" 3 baris ===
jumlah_baris = 3

for baris in range(jumlah_baris):
    baris_teks = ""
    for _ in range(1):     # inner loop: 1 kata per baris
        baris_teks += "Python "
    print(baris_teks.strip())
```

```python
# === PRAKTIKUM 5A: Mencetak "Python" 3 baris ===
jumlah_baris = 3

for baris in range(jumlah_baris):
    baris_teks = ""
    for _ in range(3):     # inner loop: 3 kata per baris
        baris_teks += "Python "
    print(baris_teks.strip())
```

```python
# === PRAKTIKUM 5B: Pola Segitiga Bintang ===
print("Pola Segitiga:")
for baris in range(1, 6):           # baris 1 sampai 5
    for kolom in range(baris):      # kolom bertambah sesuai baris
        print("*", end=" ")
    print()  # pindah baris
```

```python
# === PRAKTIKUM 5B: Pola Segitiga Bintang ===
print("Pola Segitiga:")
for baris in range(1, 6):           # baris 1 sampai 5
    for kolom in range(5, baris, -1):      # kolom bertambah sesuai baris
        print("*", end=" ")
    print()  # pindah baris
```

```python
# === PRAKTIKUM 5C: Tabel Suhu Sensor (Minggu × Hari) ===
nama_unit = ["Unit 1", "Unit 2", "Unit 3"]
suhu_harian = [
    [85.2, 87.1, 86.5, 88.0, 84.9, 85.7, 86.2],   # Unit 1
    [90.1, 91.3, 89.7, 92.0, 90.5, 91.1, 90.8],   # Unit 2
    [78.4, 79.2, 77.8, 80.1, 79.5, 78.9, 79.7],   # Unit 3
]
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Laporan Suhu Boiler (°C)")
print(f"{'Unit':<10}", end="")
for h in hari:
    print(f"{h:>6}", end="")
print(f"{'Rata2':>8}")
print("-" * 60)

for i, unit in enumerate(nama_unit):   # outer loop: per unit
    print(f"{unit:<10}", end="")
    total = 0
    for suhu in suhu_harian[i]:         # inner loop: per hari
        print(f"{suhu:>6.1f}", end="")
        total += suhu
    rata2 = total / len(suhu_harian[i])
    print(f"{rata2:>8.1f}")
```

PERTANYAAN ANALISIS

1. Berapa total iterasi yang terjadi pada Praktikum 5C (hitung outer × inner)?

    Outer loop berjalan 3 kali (satu per unit) dan setiap kali itu inner loop berjalan 7 kali (satu per hari). Jadi total iterasi inner loop adalah 3 × 7 = 21 iterasi, artinya ada 21 data suhu yang diproses. Jika outer loop ikut dihitung (3 iterasi), totalnya 24 iterasi.

2. Pada nested loop, variabel perulangan luar (i) dan dalam (suhu) memiliki scope yang berbeda. Jelaskan apa yang terjadi jika keduanya menggunakan nama variabel yang sama

    Di Python, perulangan for tidak membuat scope baru. Variabel loop luar dan dalam berada di scope yang sama (scope fungsi atau global), sehingga jika namanya sama, variabel dalam akan menimpa nilai variabel luar. Sebagai contoh, jika pada Praktikum 5C inner loop ditulis for i in suhu_harian[i]:, maka setelah inner loop selesai, i berisi angka suhu terakhir (misalnya 86.2), bukan lagi nomor unit. Baris len(suhu_harian[i]) kemudian menimbulkan TypeError karena indeks list tidak boleh berupa float. Pada kasus lain tidak muncul error, tetapi hasilnya diam-diam salah. Oleh karena itu, variabel loop luar dan dalam sebaiknya selalu diberi nama berbeda dan deskriptif, seperti i untuk unit dan suhu untuk nilai suhu.
3. Kapan nested loop menjadi tidak efisien? Apa alternatifnya dalam Python untuk pemrosesan data tabular?

    Nested loop menjadi tidak efisien ketika datanya sangat besar, karena jumlah iterasinya mengalikan ukuran kedua loop (kompleksitas O(n × m)). Jika ada 1.000 unit dengan 1.000 pembacaan, terjadi 1.000.000 iterasi, dan loop murni Python relatif lambat. Kode dengan banyak tingkat nested loop juga sulit dibaca dan dirawat. Alternatif untuk data tabular adalah library NumPy dan pandas, yang memproses seluruh array sekaligus (vektorisasi) dengan kecepatan jauh lebih tinggi. Contohnya, rata-rata per unit pada Praktikum 5C cukup ditulis:

```python
import numpy as np
print(np.mean(suhu_harian, axis=1))
```
