1. Menentukan tahun kabisat

studi kasus: sebuah tahun disebut tahun kabisat jika memenuhi salah satu dari dua kondisi berikut

a. Tahun tersebut  habis dibagi 4, tetapi tidak habis dibagi 100.

b. Tahunu tersebut habis dibagi 400

Buatlah sebuah ekspresi logika kompleks untuk menentukan apakah suatu tahun merupakan tahun kabisat.

```python
# LATIHAN 1 - MENENTUKAN TAHUN KABISAT

# Kondisi True
# Input tahun dari keyboard
tahun = int(input("Masukkan tahun: "))

# Ekspresi logika kompleks
kondisi_a = tahun % 4 == 0 and tahun % 100 != 0    # habis dibagi 4, tidak habis dibagi 100
kondisi_b = tahun % 400 == 0                       # habis dibagi 400
kabisat = kondisi_a or kondisi_b                   # salah satu terpenuhi

# Output
print("\n" + "=" * 33)
print(f"Tahun                      : {tahun}")
print(f"Habis dibagi 4 & bukan 100 : {kondisi_a}")
print(f"Habis dibagi 400           : {kondisi_b}")
print("=" * 33 + "\n")
print(f"Apakah tahun kabisat?      : {kabisat}")
```

2. Buatlah program sederhana versi anda masing-masing dengan lingkup, fungsi input, fungsi output dan opeasi aritmatika

```python
# LATIHAN 2 - KALKULATOR IMT

# INPUT
nama = input("Masukkan nama Anda: ")
berat = float(input("Masukkan berat badan (kg): "))
tinggi_cm = float(input("Masukkan tinggi badan (cm): "))

# PROSES
tinggi_m = tinggi_cm / 100
imt = berat / (tinggi_m ** 2)

# OUTPUT
print("\n=== HASIL PERHITUNGAN IMT ===")
print(f"Nama         : {nama}")
print(f"Berat Badan  : {berat} kg")
print(f"Tinggi Badan : {tinggi_cm} cm ({tinggi_m} m)")
print(f"Nilai IMT    : {imt:.2f}")

# Kategori sederhana (opsional)
if imt < 18.5:
    kategori = "Kurus"
elif imt < 24:
    kategori = "Normal"
elif imt < 28:
    kategori = "Gemuk"
else:
    kategori = "Obesitas"

print(f"Kategori     : {kategori}")
```
