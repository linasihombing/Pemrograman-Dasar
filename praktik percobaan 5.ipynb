Jajan di Kantin

Rani membeli 3 gorengan seharga Rp 2.500 per buah dan 2 es teh seharga Rp 4.000 per gelas. Rani membayar dengan uang Rp 20.000.

Tentukan:

Total belanja. Uang kembalian. Apakah uang Rani cukup? (True/False

1.   Hitung harga gorengan: jumlah × harga satuan.
2.   Jumlahkan keduanya menjadi total.
3.   Kembalian = uang dibayar dikurangi total.
4.   Hitung harga es teh: jumlah × harga satuan.
5.   Bandingkan uang dengan total memakai >=.



```python
# Jajan di Kantin

# variabel
jumlah_gorengan = int (input ("masukan jumlah goregan yang dibeli: "))
harga_gorengan = int (input ("masukan harga satuan gorengan: "))

jumlah_es_teh = int (input ("masukan jumlah es teh yang dibeli: "))
harga_es_teh = int (input ("masukan harga satuan es teh: "))

uang_dibayar = int (input ("masukan jumlah uang: "))

# Hitung harga gorengan
harga_total_gorengan = jumlah_gorengan * harga_gorengan

# Hitung harga es teh
harga_total_es_teh = jumlah_es_teh * harga_es_teh

# Jumlahkan menjadi total belanja
total_belanja = harga_total_gorengan + harga_total_es_teh

# Hitung uang kembalian
uang_kembalian = uang_dibayar - total_belanja

# Apakah uang Rani cukup
uang_cukup = uang_dibayar >= total_belanja

# Output
print ()
print("Harga total gorengan :", harga_total_gorengan)
print("Harga total es teh   :", harga_total_es_teh)
print("Total belanja        :", total_belanja)
print("Uang kembalian       :", uang_kembalian)
print("Apakah uang cukup?   :", uang_cukup)
```

# **Durasi Film Maraton**

Dina menonton film maraton dengan total durasi 275 menit.

Tentukan:

Berapa jam penuh dan berapa menit sisa (dalam format "... jam ... menit"). Total durasi dalam jam desimal. Apakah durasi tersebut lebih dari 3 jam dan kurang dari 5 jam? (True/False)

Todo
1. Jam penuh = total menit dibagi 60 tanpa pecahan (//).
2. Menit sisa = sisa pembagian total menit dengan 60 (%).
3. Jam desimal = total menit dibagi 60 dengan pecahan (/).
4. Cek rentang dengan > dan < yang digabung and.

```python
# Durasi Film Maraton

total_menit = 275

# Hitung jam penuh
jam_penuh = total_menit // 60

# Hitung menit sisa
menit_sisa = total_menit % 60

# Hitung durasi dalam jam desimal
jam_desimal = total_menit / 60

# Cek apakah lebih dari 3 jam dan kurang dari 5 jam
dalam_rentang = (jam_desimal > 3) and (jam_desimal < 5)

# Output
print(f"Durasi: {jam_penuh} jam {menit_sisa} menit")
print(f"Durasi dalam jam desimal: {jam_desimal}")
print(f"Lebih dari 3 jam dan kurang dari 5 jam: {dalam_rentang}")
```
