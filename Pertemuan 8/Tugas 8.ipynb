## **Aktivitas 6F**

Lanjutkan tabel secara manual untuk dua elemen terakhir (75 dan 70). Berapa total akhir dan rata-ratanya? Berapa skor yang bernilai 75 atau lebih? Lalu jalankan blok berikut.

```python
skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print()
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

## **Aktivitas 6G**

Ubah blok di atas menjadi rata-rata bergerak 5 hari (mulai perulangan di indeks 4, slice suhu[i - 4 : i + 1], bagi dengan 5). Berapa rata-rata yang Anda dapat dari 7 nilai? Mengapa?

```python
suhu = [28, 31, 30, 33, 29, 27, 32]

rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

## **Aktivitas 6H**

Jelaskan penurunannya

Energi hari ke-17 turun lebih dari 40%. Dalam kelompok, tuliskan tiga alasan operasional yang mungkin menjelaskannya dan data apa yang akan Anda periksa untuk memastikan masing-masing. (Petunjuk: lihat jam_operasi[16] di bagian data 30 hari, dan blok hari kerja di bawah.)

```python
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# Perubahan harian: mulai dari indeks 1 karena dibandingkan dengan elemen sebelumnya
for i in range(1, len(energi)):
    perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100
    if perubahan < -20:
        print(f"Hari {i + 1}: {perubahan:+.1f}% dibanding hari sebelumnya, perlu diselidiki")

# Rata-rata bergerak 7 hari
rata_7 = []
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7.append(sum(jendela) / 7)

print(f"{len(rata_7)} nilai rata-rata bergerak (data {len(energi)} hari)")
print(f"Terendah {min(rata_7):,.0f} MWh/hari, tertinggi {max(rata_7):,.0f} MWh/hari")
```

## **Aktivitas 6I**

Array Anda sendiri

Buat list suhu berisi 7 suhu harian (bebas). Lalu: (1) tampilkan 3 hari pertama dan 2 hari terakhir dengan slicing; (2) tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop; (3) tampilkan versi urutnya dengan sorted tanpa mengubah suhu; (4) hitung rata-rata dan cari suhu tertinggi beserta hari ke berapa; (5) buat list baru berisi suhu di atas rata-rata.

```python
suhu = [29, 31, 30, 33, 28, 32, 30]   # 7 suhu harian (°C), bebas

# (1) 'Slicing: 3 hari pertama dan 2 hari terakhir
print("Jawaban 1")
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])
print()

# (2) append lalu pop
suhu.append(34)
print("Jawaban 2")
print("Setelah append:", suhu, "| panjang:", len(suhu))
dikeluarkan = suhu.pop()
print("Dikeluarkan:", dikeluarkan, "| panjang:", len(suhu))
print()

# (3) Versi urut tanpa mengubah suhu
urut = sorted(suhu)
print("Jawaban 3")
print("Urut  :", urut)
print("Asli  :", suhu)
print()

# (4) Rata-rata dan suhu tertinggi beserta harinya
rata_rata = sum(suhu) / len(suhu)
i_maks = suhu.index(max(suhu))
print("jawaban 4")
print(f"Rata-rata: {rata_rata:.1f} °C")
print(f"Tertinggi: {suhu[i_maks]} °C pada hari ke-{i_maks + 1}")
print()

# (5) List baru: suhu di atas rata-rata
di_atas = [t for t in suhu if t > rata_rata]
print("Jawaban 5")
print("Di atas rata-rata:", di_atas)
print()
```

## **Pemanasan**
Lima latihan kecil (data kecil)

Pakai skor = [80, 65, 90, 75, 70, 85]. Tulis kode yang (1) mencetak skor pertama dan skor terakhir (pakai indeks negatif untuk yang terakhir), (2) mencetak rata-rata, (3) menghitung berapa skor di atas rata-rata, (4) mencetak posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1), dan (5) membuat list baru dengan setiap skor ditambah 5.

```python
skor = [80, 65, 90, 75, 70, 85]

# (1) Skor pertama dan terakhir (indeks negatif untuk yang terakhir)
print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

# (2) Rata-rata
rata = sum(skor) / len(skor)
print("Rata-rata:", rata)

# (3) Banyaknya skor di atas rata-rata
jumlah_atas = 0
for nilai in skor:
    if nilai > rata:
        jumlah_atas = jumlah_atas + 1
print("Skor di atas rata-rata:", jumlah_atas)

# (4) Posisi skor terendah sebagai nomor mahasiswa
i_min = skor.index(min(skor))
print("Skor terendah:", skor[i_min], "milik mahasiswa nomor", i_min + 1)

# (5) List baru: setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)
print("Skor + 5:", skor_baru)
```

## **Latihan praktikum**

Dengan list energi 30 hari (salin dari blok kode di atas), tulis program yang mencetak: total dan rata-rata energi; tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya); jumlah hari di bawah target yang Anda masukkan; setiap hari yang rata-rata bergerak 7 harinya turun di bawah 97% rata-rata bulanan; serta total energi per minggu (minggu 1–4, ditambah sisa hari).

```python
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# 1. Total dan rata-rata
total = sum(energi)
rata = total / len(energi)
print("=== LAPORAN PRODUKSI NOVEMBER 2026 ===")
print(f"Total energi : {total:,} MWh")
print(f"Rata-rata    : {rata:,.1f} MWh/hari")

# 2. Tiga tertinggi dan tiga terendah (urutkan INDEKS berdasarkan nilainya)
urut_indeks = sorted(range(len(energi)), key=lambda i: energi[i])
print("\nTiga hari terendah:")
for i in urut_indeks[:3]:
    print(f"  Hari ke-{i + 1}: {energi[i]:,} MWh")
print("Tiga hari tertinggi:")
for i in urut_indeks[::-1][:3]:
    print(f"  Hari ke-{i + 1}: {energi[i]:,} MWh")

# 3. Jumlah hari di bawah target
target = int(input("\nMasukkan target harian (MWh): "))
hari_kurang = [i + 1 for i in range(len(energi)) if energi[i] < target]
print(f"Hari di bawah target {target:,}: {len(hari_kurang)} hari -> {hari_kurang}")

# 4. Rata-rata bergerak 7 hari di bawah 97% rata-rata bulanan
batas = 0.97 * rata
print(f"\nBatas 97% rata-rata bulanan: {batas:,.1f} MWh/hari")
ada = False
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7
    if rata_7 < batas:
        print(f"  Jendela berakhir hari ke-{i + 1}: {rata_7:,.1f} MWh/hari")
        ada = True
if not ada:
    print("  Tidak ada.")

# 5. Total per minggu
print("\nTotal per minggu:")
for m in range(4):
    awal = m * 7
    print(f"  Minggu {m + 1} (hari {awal + 1}-{awal + 7}): {sum(energi[awal:awal + 7]):,} MWh")
sisa = energi[28:]
print(f"  Sisa (hari 29-30): {sum(sisa):,} MWh")
```
