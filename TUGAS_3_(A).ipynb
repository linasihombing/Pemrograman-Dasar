**JAWABAN SOAL ESAI NO.2**

Perbedaan mengenai operator relasional dan operator logika adalah operator relasional (>, <, >=, <=, ==, !=) digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean (True atau False). Fungsinya dapat diterapkan untuk menjawab pertanyaan seperti “apakah nilai A lebih besar dari nilai B?”, “apakah kedua nilai itu sama?”, dan sejenisnya.

Namun, Operator logika (and, or, not) digunakan untuk menggabungkan atau membalik nilai boolean. Operator ini tidak membandingkan angka secara langsung, tetapi bekerja pada hasil dari operasi boolean (misalnya hasil operator relasional).

```python
#contoh operator rasional sebagai berikut:

#penerapan pada tanda (>)
print("1. Apakah nilai 5 lebih besar daripada nilai 3?")
print(5 > 3)   # Output: True
#penerapan pada tanda (<)
print("2. Apakah nilai 5 lebih kecil daripada nilai 3?")
print(5 < 3)   # Output: False
#penerapan pada tanda (>=)
print("3. Apakah nilai 5 lebih besar atau sama dengan nilai 3?")
print(5 >= 3)  # Output: True
print("4. Apakah nilai 3 lebih besar atau sama dengan dari nilai 5?")
print(3 >= 5)  # Output: False
#penerapan pada tanda (<=)
print("5. Apakah nilai 5 lebih kecil atau sama dengan nilai 3?")
print(5 <= 3)  # Output: False
print("6. Apakah nilai 3 lebih kecil atau sama dengan dari nilai 5?")
print(3 <= 5)  # Output: True
#penerapan pada tanda (==)
print("7. Apakah nilai 10 sama dengan nilai 10?")
print(10 == 10) # Output: True
print("8. Apakah nilai 10 sama dengan nilai 5?")
print(10 == 5)  # Output: False
#penerapan pada tanda (!=)
print("9. Apakah nilai 10 tidak sama dengan nilai 10?")
print(10 != 10) # Output: False
print("10. Apakah nilai 10 tidak sama dengan nilai 5?")
print(10 != 5)  # Output: True
```

```python
# CONTOH OPERATOR LOGIKA

# PENRAPAN PADA TANDA "AND"
# Contoh 1: Kedua kondisi True
print("1. Apakah nilai 5 lebih besar dari 2 DAN nilai 10 lebih besar dari 3?")
print((5 > 2) and (10 > 3))  #jika kedua nilai adalah True and True maka hasilnya akan True
# Contoh 2: Salah satu kondisi False
print("2. Apakah nilai 5 lebih besar dari 2 DAN nilai 10 lebih kecil dari 3?")
print((5 > 2) and (10 < 3))  #jiak kedua nilai adalah True and False maka hasilnya akan False
# PENRAPAN PADA TANDA "OR"
# Contoh 3: Salah satu kondisi True
print("3. Apakah nilai 5 lebih kecil dari 2 ATAU nilai 10 lebih kecil dari 3?")
print((5 > 2) or (10 < 3))   #jika kedua nilai adalah True and True maka hasilnya akan True
# Contoh 4: Kedua kondisi False
print("4. Apakah nilai 5 lebih kecil dari 2 ATAU nilai 10 lebih kecil dari 3?")
print((5 < 2) or (10 < 3))   #jika kedua nilai adalah False and False maka hasilnya akan False
```

```python
# CONTOH SITUASI NYATA TERHADAP OPERATOR RASIONAL DAN OPERATOR LOGISTIK

# Sistem Seleksi Beasiswa

# Input data mahasiswa
nama = "Maulina"
ipk = 3.95
usia = 19
status_cuti = False  # False = tidak cuti, True = sedang cuti

# Kriteria beasiswa:
syarat_IPK  = ipk >= 3.50
syarat_usia = usia <= 22
syarat_cuti = status_cuti

# Menggunakan operator relasional dan logika
layak_beasiswa = (syarat_IPK) and (syarat_usia) and (not syarat_cuti)

# Menampilkan hasil
print("SELEKSI BEASISWA")
print("-----------------")
print()
print("Nama        :", nama)
print("IPK         :", ipk)
print("Usia        :", usia)
print("Sedang cuti :", status_cuti)
print()

if layak_beasiswa:
    print("Hasil: MAULINA LAYAK MENDAPAT BEASISWA.")
else:
    print("Hasil: MAULINA TIDAK LAYAK MENDAPAT BEASISWA.")
```

```python
#Diskon E-commerce

#Data transaksi
total_belanja = 650000      # dalam rupiah
anggota_premium = False     # True = anggota premium, False = bukan
sedang_promo = False        # True = sedang ada promo, False = tidak ada promo

# Aturan diskon:
syarat_belanja = total_belanja >= 500000
syarat_premium = anggota_premium
syarat_promo = sedang_promo

diskon_diberikan = (syarat_belanja and syarat_premium) or syarat_promo

# Menampilkan hasil
print("DISKON E-COMMERCE")
print("------------------")
print()
print("Total belanja   :", total_belanja)
print("Anggota premium :", anggota_premium)
print("Sedang promo    :", sedang_promo)
print()

if diskon_diberikan:
    print("Hasil: TRANSAKSI INI MENDAPAT DISKON.")
else:
    print("Hasil: TRANSAKSI INI TIDAK MENDAPAT DISKON.")
```

**JAWABAN SOAL ESAI NO.3**

Kasus tersebut adalah bug perhitungan akibat operator precedence dalam Python. Programmer ingin menghitung rata-rata tiga nilai dengan rumus (nilai 1 + nilai 2 + nilai 3)/3
etapi kode yang ditulis adalah scores[0] + scores[1] + scores[2] / 3. Dalam Python, operator pembagian / memiliki prioritas lebih tinggi daripada penjumlahan +, sehingga Python menghitung scores[2] / 3 terlebih dahulu, kemudian hasilnya ditambahkan dengan dua nilai sebelumnya. Untuk input 15, 30, 45, Python sebenarnya menghitung 15 + 30 + (45 / 3), yaitu 15 + 30 + 15 = 60.0, bukan (15 + 30 + 45) / 3 = 30.0. Program tidak menampilkan error karena penulisannya secara sintaks sudah benar, tetapi hasil perhitungannya salah; inilah yang disebut logic bug atau bug logika. Kasus serupa dibahas dalam pertanyaan programmer di Stack Overflow tentang kebingungan perhitungan rata-rata dasar di Python.

Berikut ini saya coba terapkan bug atau kesalahan perhitungan yang terjadi akibat salah memahami operator precedence agar dapat menjadi perbandingan sebagai evaluasi dan juga menyertakan penerapan kode dengan cara yang benar dibawah ini:

```python
# KODE SALAH: contoh bug operator precedence

print("Program menghitung rata-rata dari 3 nilai")

nilai1 = 15
nilai2 = 30
nilai3 = 45

print()
print("Nilai 1 =", nilai1)
print("Nilai 2 =", nilai2)
print("Nilai 3 =", nilai3)
print()

# Ekspresi yang ditulis:
print("nilai1 + nilai2 + nilai3 / 3")

# Python membacanya sebagai:
# ("nilai1 + nilai2 + (nilai3 / 3)")

# Kesalahan akibat kode: Python menghitung nilai3 / 3 lebih dulu
rata_rata_salah = nilai1 + nilai2 + nilai3 / 3

print()
print("Hasil kode salah =", rata_rata_salah)
```

```python
# KODE BENAR: gunakan tanda kurung

nilai1 = 15
nilai2 = 30
nilai3 = 45

print()
print("Nilai 1 =", nilai1)
print("Nilai 2 =", nilai2)
print("Nilai 3 =", nilai3)
print()

# Benar: ketiga nilai dijumlahkan dahulu, baru dibagi 3
rata_rata_benar = (nilai1 + nilai2 + nilai3) / 3

# Ekspresi yang benar
print("kode yang benar adalah:", "(nilai1 + nilai2 + nilai3) / 3")

print()
print("Hasil kode benar =", rata_rata_benar)
```
