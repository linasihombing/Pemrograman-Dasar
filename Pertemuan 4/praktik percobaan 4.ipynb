```python
# PRAKTIK INPUT DAN OUTPUT (2) - PRAKTIK 1

# Berikut ini beberapa contoh penggunaan karakter unutk fuungsi output
nama_pelanggan = "Hendi"
menu           = "Mi ayam bakso"
jumlah_porsi   = 2
harga_satuan   = 15000.50
kode_transaksi = 255
kode_antrian   = 64

# 1. Menggunakan %s untuk String
print("Nama Pelanggan      : %s" % nama_pelanggan)
print("Pesanan             : %s" % menu)

# 2. Menggunakan %d untuk Integer
print("Jumlah Porsi        : %d porsi" % jumlah_porsi)

# 3. Menggunakan %f untuk Float (menggunakan %.2f untuk membatasi 2 angka di belakang koma)
print("Harga Satuan        : Rp %.2f" % harga_satuan)

# 4. Menggunakan %x untuk Hexadesimal
print("Nilai hex dari 255  : %x" % kode_transaksi)

# 5. Menggunakan %o unutk oktal
print("Nilai oktal dari 64 : %o" % kode_antrian)
```

```python
# PRAKTIK EKSPRESI OPERATOR ARITMATIKA (2) - PRAKTIK 1

# MENGHITUNG NILAI AKHIR MAHASISWA
# 1. Input 5 komponen nilai dari keyboard
print("=== INPUT NILAI MAHASISWA ===")
nilai_tugas   = float(input("Masukkan nilai Tugas   : "))
nilai_kuis    = float(input("Masukkan nilai Kuis    : "))
nilai_uts     = float(input("Masukkan nilai UTS     : "))
nilai_uas     = float(input("Masukkan nilai UAS     : "))
nilai_praktik = float(input("Masukkan nilai Praktik : "))

# 2. Hitung nilai akhir dengan bobot tertentu
#    Misal: Tugas 20%, Kuis 10%, UTS 30%, UAS 30%, Praktik 10%
nilai_akhir = ( 0.20 * nilai_tugas + 0.10 * nilai_kuis + 0.30 * nilai_uts + 0.30 * nilai_uas + 0.10 * nilai_praktik)

# 3. Input standar kelulusan dari keyboard
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

# 4. Tentukan status lulus / remedial
apakah_lulus = nilai_akhir >= standar_kelulusan
perlu_remedial = nilai_akhir < standar_kelulusan

# 5. Tampilkan hasil
print("\n=== HASIL KELULUSAN ===")
print(f"Nilai Akhir Mahasiswa: {nilai_akhir}")
print(f"Standar Kelulusan    : {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial?         {perlu_remedial}")
```

```python
# PRAKTIK EKSPRESI OPERATOR ARITMATIKA (3) - PRAKTIK 1

# MENGHITUNG HASIL BELANJA (DISKON 35% UNTUK BELANJA > 1.000.000)
# kondisi mendapatkan diskon
belanja = float(input("Masukkan total belanja     = "))
diskon = (0, 0.35 * belanja) [belanja > 1000000]
total_harga = belanja - diskon

print(f"Total belanja              = {belanja:.2f}")
print(f"Diskon                     = {diskon:.2f}")
print("Total harga setelah diskon =", total_harga)
```
